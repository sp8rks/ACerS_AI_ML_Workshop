```python
# sys = info about Python itself; importlib = load a library by its name (given as text)
import sys, importlib
from pathlib import Path  # Path = an easy way to work with file and folder locations

# Print the Python version; the "a if condition else b" form picks a tick or a cross
print(f"Python {sys.version.split()[0]}", "✅" if sys.version_info >= (3, 10) else "❌ need Python ≥ 3.10")
# Loop over (library name, minimum version) pairs; each pair is "unpacked" into pkg and minimum
for pkg, minimum in [("numpy", "1.26"), ("pandas", "2.1"), ("matplotlib", "3.10"),
                     ("sklearn", "1.4"), ("scipy", "1.11"), ("anthropic", "1.0")]:
    # try/except: attempt the code below, and jump to "except" if the library is missing
    try:
        mod = importlib.import_module(pkg)  # load the library (same as "import numpy" etc.)
        ver = mod.__version__  # the installed version as text, e.g. "1.26.4"
        # Turn "1.26.4" into the numbers (1, 26) and compare them with the minimum version
        ok = tuple(int(x) for x in ver.split(".")[:2]) >= tuple(int(x) for x in minimum.split("."))
        # f-string: values in {} are filled in; ":12s" pads the text to 12 characters wide
        print(f"{pkg:12s} {ver:10s}", "✅" if ok else f"⚠️  please upgrade to ≥ {minimum}")
    except ImportError:  # runs only if the library is not installed
        note = "(only needed for Session 6)" if pkg == "anthropic" else ""  # extra hint text
        print(f"{pkg:12s} {'missing':10s} ❌ {note}")

root = Path.cwd().resolve()  # start from the folder this notebook is running in
# Walk up one folder at a time until we find the workshop_utils folder (or hit the top)
while not (root / "workshop_utils").exists() and root != root.parent:
    root = root.parent  # move up to the parent folder
sys.path.insert(0, str(root))  # tell Python to look in that folder when importing
try:
    # The course's helper code: turns chemical formulas into numeric features
    from workshop_utils import generate_features, load_dataset
    # Featurize the first 50 formulas as a quick test; "X, *_" keeps only the first result
    X, *_ = generate_features(load_dataset("bulk_modulus").head(50), verbose=False)
    print(f"workshop_utils ✅  (featurized {len(X)} test formulas)")  # len(X) = number of rows
except Exception as e:  # catch any error and show its message instead of crashing
    print("workshop_utils ❌", e)
```
